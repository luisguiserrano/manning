# Gini index calculation

```python
import numpy as np
```

```python
elements = ['A', 'A', 'A', 'C', 'B', 'C']
def counts(elements):
    classes = {}
    for element in elements:
        if element in classes:
            classes[element] += 1
        else:
            classes[element] = 1
    return [classes[e] for e in classes]
counts(elements)
```

```python
def gini(elements):
    cts = counts(elements)
    n = sum(cts)
    return 1 - sum([p_i**2/n**2 for p_i in cts])
gini(elements)
```

```python
def entropy(elements):
    if len(elements)==0:
        return 0
    cts = counts(elements)
    n = sum(cts)
    props = 1/n*np.array(cts)
    return float(-np.dot(np.log2(props), props))
entropy(elements)
```

```python
for i in range(len(elements)+1):
    print("************")
    left = elements[:i]
    right = elements[i:]
    print(left, right)
    weighted_gini = 1/len(elements)*(gini(left)*len(left) + gini(right)*len(right))
    print("Weighted Gini:", weighted_gini)
    weighted_entropy = 1/len(elements)*(entropy(left)*len(left) + entropy(right)*len(right))
    print("Weighted entropy:", weighted_entropy)
```

# Worked examples from the chapter

```python
set_1 = ['red']*8 + ['blue']*2
set_2 = ['red']*4 + ['blue']*3 + ['yellow']*2 + ['green']
print(round(gini(set_1), 3), round(gini(set_2), 3))        # 0.32 0.7
print(round(entropy(set_1), 3), round(entropy(set_2), 3))  # 0.722 1.846
print(f'{0.8**8 * 0.2**2:.10f}')                            # 0.0067108864
print(f'{0.4**4 * 0.3**3 * 0.2**2 * 0.1:.10f}')            # 0.0000027648
```

```python
# Splitting the app dataset by platform and by age (figures 9.12 and 9.13)
platform = (gini(['A','C','C']) + gini(['A','A','B']))/2
age = (gini(['A','A','A']) + gini(['B','C','C']))/2
print(round(platform, 3), round(age, 3))  # 0.444 0.222
platform = (entropy(['A','C','C']) + entropy(['A','A','B']))/2
age = (entropy(['A','A','A']) + entropy(['B','C','C']))/2
print(round(platform, 3), round(age, 3))  # 0.918 0.459
```

```python
# Weighted averages for a split of 8 points into 6 and 2 (figure 9.14)
left = ['black']*2 + ['gray']*4
right = ['black']*2
print(round(6/8*gini(left) + 2/8*gini(right), 3))        # 0.333
print(round(6/8*entropy(left) + 2/8*entropy(right), 3))  # 0.689
```
